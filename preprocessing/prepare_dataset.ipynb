```python
import json
import pandas as pd
import random
```

```python
corpus_df = pd.read_csv("data/full_corpus.csv")
train_df = pd.read_csv("data/train.csv")
test_df = pd.read_csv("data/public_test.csv")
```

```python
train_df
```

```python
cids=corpus_df["cid"].to_list()
```

```python
cid_text_map = dict(zip(corpus_df['cid'], corpus_df['text']))
```

```python
random.randint(0, 5)
```

```python
dataset_triplet = []
total_len = 0
```

```python
for i, r in train_df.iterrows():
    if isinstance(r["cid"], str):
        list_cid = r["cid"][1:-1].split()
        list_cid = [int(cid) for cid in list_cid]
        pos_len = len(list_cid)
        total_len += pos_len
        neg_len = 0
        while neg_len < pos_len:
            rnd_idx = random.randint(0, len(cids) - 1)
            if cids[rnd_idx] not in list_cid:
                dataset_triplet.append({"qid": r["qid"], 
                                        "question": r["question"], 
                                        "positive": 
                                            {"cid": list_cid[neg_len], 
                                             "text":cid_text_map[list_cid[neg_len]]},
                                        "negative":
                                            {"cid": cids[rnd_idx],
                                             "text": cid_text_map[cids[rnd_idx]],}
                                            })
                neg_len += 1
```
